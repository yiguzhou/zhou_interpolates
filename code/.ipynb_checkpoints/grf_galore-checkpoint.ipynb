```python
import gstools as gs

gstools_mdls = {'spherical': gs.covmodel.Spherical, 
                'exponential': gs.covmodel.Exponential,
                'gaussian': gs.covmodel.Gaussian,
                'cubic': gs.covmodel.Cubic,
                'stable': gs.covmodel.Stable,
                'matern': gs.covmodel.Matern,
               'hole': gs.covmodel.JBessel}
```

```python
import joblib

sphere_mesh = joblib.load(projdir + '/data/fsLR-4k_sphere_grf.pkl')
```

```python
import copy
from interpmodules import helpers
import pyvista as pv
import random

ranges=[10, 25, 50, 75, 100]

pl = pv.Plotter(shape = (len(ranges), len(gstools_mdls)), window_size = (len(gstools_mdls)*200, len(ranges)*200))

Instances = {}

for ri,ran in enumerate(ranges):

    instances = {}
    
    for mi, model in enumerate(gstools_mdls):
        
        l = helpers._convert_to_length(ran)
        
        mdl = gstools_mdls[model](dim = 3, len_scale=l, var=1.0)
        srf = gs.SRF(mdl)

        seed = random.randint(0,100000)
        random_map = srf.mesh(sphere_mesh, points='points', seed=seed)
    
        instances[model] = random_map

        pl.subplot(ri,mi)
        if ri==0:
            pl.add_text(model, position="upper_edge", font = 'courier', font_size=10)
        if mi==0:
            pl.add_text(f"{str(ran)}mm", position="left_edge", font = 'courier', orientation = 90, font_size=8)
        tmp_mesh = copy.deepcopy(sphere_mesh)
        pl.add_mesh(tmp_mesh, scalars=instances[model], cmap = bpr, **mesh_aes)
        
        pl.camera_position='yz'

    Instances[ran] = pd.DataFrame(instances)

pl.show(jupyter_backend='static')
```

```python
fig, axs = plt.subplots(1, len(ranges), figsize=(len(ranges)*4,4), sharex=True, sharey=True)

for ri, ran in enumerate(ranges):

    instances = Instances[ran]

    crosscorr = np.zeros((len(gstools_mdls), len(gstools_mdls)))
    
    for mi, model1 in enumerate(gstools_mdls):
        for mj, model2 in enumerate(gstools_mdls):
                
            crosscorr[mi, mj] = np.corrcoef(instances[model1], instances[model2])[1,0]
            if mi == mj: crosscorr[mi, mj] = np.nan

    sns.heatmap(crosscorr, square=True, vmin=0, vmax=1, cmap = 'Reds', linewidth = 1, linecolor = 'w',
                xticklabels=gstools_mdls.keys(), yticklabels=gstools_mdls.keys(), cbar = False, ax=axs[ri])
    if ri == len(ranges) - 1:
        sns.heatmap(crosscorr, square=True, vmin=0, vmax=1, cmap = 'Reds', linewidth = 1, linecolor = 'w',
                xticklabels=gstools_mdls.keys(), yticklabels=gstools_mdls.keys(), ax=axs[ri])
        
    axs[ri].collections[0].cmap.set_bad('lightgray')
    axs[ri].set_title(f'{str(ran)}mm')
    
plt.tight_layout()
plt.show()
```

# look at variogram

```python
from interpmodules import helpers

X0=kc[:,:,0]
Y0=train
Dist = helpers.internal_distance_matrix(sphere_mesh.points, 'euclidean')
h, ev, N = helpers.get_empirical_variogram_fast(x=Y0,X=X0,D=Dist,nh=25,h_bounds='auto', method='euclidean')
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(4,2))
plt.scatter(h,ev,c='dimgray')
plt.show()
```

```python
import numpy as np
# starting point for parameter values

def init_parameters(mapvals, Dist, h, ev):

    nh = len(h)
    p0 = {
          3: [np.max(h), np.nanmax(ev), np.nanmax(ev)/nh],
          4: [np.max(h), np.nanmax(ev), np.nanmax(ev)/nh, ev[0]]
         }
    # upper bound lower bound
    bounds = {3: ([0.00001]*3, [np.max(Dist), semivariance(np.nanmax(mapvals), np.nanmin(mapvals)),
                        semivariance(np.nanmax(mapvals), np.nanmin(mapvals))]),
    4: ([0.00001]*4, [np.max(Dist), semivariance(np.nanmax(mapvals), np.nanmin(mapvals)),
                        semivariance(np.nanmax(mapvals), np.nanmin(mapvals)), 100])}

    return p0, bounds
```

```python
def semivariance(y1, y2):
    return .5 * (y1 - y2)**2
```

```python
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from sklearn.metrics import r2_score
from interpmodules import metrics
from scipy.stats import zscore

# Fitting
mapvals = dt

Cof = {}
Cov = {}
gof = {}

fig, axs = plt.subplots(1, len(gstools_mdls), figsize = (14,2), sharey=True)

ev[np.where(np.isnan(ev))] = 0

model_scores = {}

for ind, model in enumerate(gstools_mdls):

    cof = None; cov = None

    axs[ind].scatter(h, ev, marker = "+", color = 'dimgray')
    
    p0, bounds = init_parameters(mapvals, Dist, h, ev)

    mdl = gstools_mdls[model](dim=3)
    
    if model in ['stable', 'matern']:
            
        in_bounds = {"len_scale": [bounds[4][0][0], bounds[4][1][0]],\
                 "var": [bounds[4][0][1], bounds[4][1][1]],\
                 "nugget": [bounds[4][0][2], bounds[4][1][2]],\
                 "smoothness": [bounds[4][0][3], bounds[4][1][3]]
                }

        if model == 'stable':
            in_bounds["alpha"] = [bounds[4][0][3], 2]
            del in_bounds["smoothness"]
            
        if model == 'matern':
            in_bounds["nu"] = [bounds[4][0][3], 20]
            del in_bounds["smoothness"]

        
        mdl.set_arg_bounds(**in_bounds)
        
        cof, cov = mdl.fit_variogram(h,ev)
    
    else:
        in_bounds = {"len_scale": [bounds[3][0][0], bounds[3][1][0]],\
                 "var": [bounds[3][0][1], bounds[3][1][1]],\
                 "nugget": [bounds[3][0][2], bounds[3][1][2]]
                }
        
        mdl.set_arg_bounds(**in_bounds)
        cof, cov = mdl.fit_variogram(h,ev)

    Cof[model] = cof
    Cov[model] = cov
    
    v_pred = mdl.variogram(h)
    x_s = np.linspace(h.min(), h.max())
    axs[ind].plot(h, v_pred, color = 'b')
    mae = np.mean(metrics.metric(v_pred, ev, 'residual'))
    r2 = r2_score(ev, v_pred, force_finite=False)
    gof[model] = r2
    
    axs[ind].text(x=0.2*np.nanmax(h), y=0.2*np.nanmax(ev),s= model + "\n" + r'MAE=' + '{0:.2f}'.format(mae))

model_scores[ran] = gof
```

```python
from pykrige.ok3d import OrdinaryKriging3D

x1=uc[:,:,0]

preds = dict.fromkeys(gstools_mdls.keys())

for ind, model in enumerate(gstools_mdls):
    fitted_vario = gstools_mdls[model](dim=3, **Cof[model])

    kriger = OrdinaryKriging3D(x=X0[:,0], y=X0[:,1], z=X0[:,2], val = Y0,\
                                   variogram_model = fitted_vario, \
                                exact_values = True)
    y1, error = kriger.execute(style = 'points', xpoints = x1[:,0], ypoints = x1[:,1], zpoints = x1[:, 2])

    full_map = np.zeros(sphere_mesh.n_points)
    full_map[kv[:,0].astype(int)] = mapvals[kv[:,0].astype(int)]
    full_map[uv[:,0].astype(int)] = y1

    preds[model] = full_map
    print("done model: " + model + " with coefficients " + str(Cof[model]))
```

# do all the models by looping

```python
Dist = helpers.internal_distance_matrix(sphere_mesh.points, 'euclidean')
```

```python
percentages = [5, 10, 25, 50]
```

```python
for ran in ranges:
    NRMSE = {}
    
    for pct in percentages:
    
        sampl = joblib.load(projdir + f'/data/sampling/fsLR-4k_sphere_samp-fibonacci_pct-{str(pct)}.pkl')
        kc, kv, uc, uv = sampl[1]
    
        
        Nrmse = np.ones((len(gstools_mdls), len(gstools_mdls)))
        
        for mi, model in enumerate(gstools_mdls.keys()):
        
            dt = zscore( Instances[ran][model].to_numpy() )
            train = dt[kv.flatten().astype(int)]
            test = dt[uv.flatten().astype(int)]
                
            X0=kc[:,:,0]
            Y0=train
            
            h, ev, N = helpers.get_empirical_variogram_fast(x=Y0,X=X0,D=Dist,nh=25,h_bounds='auto', method='euclidean')
        
            # Fitting
            mapvals = dt
            
            Cof = {}
            Cov = {}
            gof = {}
                    
            ev[np.where(np.isnan(ev))] = 0
            
            model_scores = {}
            
            for ind, model in enumerate(gstools_mdls):
            
                cof = None; cov = None
                        
                p0, bounds = init_parameters(mapvals, Dist, h, ev)
            
                mdl = gstools_mdls[model](dim=3)
                
                if model in ['stable', 'matern']:
                        
                    in_bounds = {"len_scale": [bounds[4][0][0], bounds[4][1][0]],\
                             "var": [bounds[4][0][1], bounds[4][1][1]],\
                             "nugget": [bounds[4][0][2], bounds[4][1][2]],\
                             "smoothness": [bounds[4][0][3], bounds[4][1][3]]
                            }
            
                    if model == 'stable':
                        in_bounds["alpha"] = [bounds[4][0][3], 2]
                        del in_bounds["smoothness"]
                        
                    if model == 'matern':
                        in_bounds["nu"] = [bounds[4][0][3], 20]
                        del in_bounds["smoothness"]
            
                    
                    mdl.set_arg_bounds(**in_bounds)
                    
                    cof, cov = mdl.fit_variogram(h,ev)
                
                else:
                    in_bounds = {"len_scale": [bounds[3][0][0], bounds[3][1][0]],\
                             "var": [bounds[3][0][1], bounds[3][1][1]],\
                             "nugget": [bounds[3][0][2], bounds[3][1][2]]
                            }
                    
                    mdl.set_arg_bounds(**in_bounds)
                    cof, cov = mdl.fit_variogram(h,ev)
            
                Cof[model] = cof
                Cov[model] = cov
                
                v_pred = mdl.variogram(h)
                x_s = np.linspace(h.min(), h.max())
                mae = np.mean(metrics.metric(v_pred, ev, 'residual'))
                r2 = r2_score(ev, v_pred, force_finite=False)
                gof[model] = r2
                        
            model_scores[ran] = gof
        
            x1=uc[:,:,0]
        
            preds = dict.fromkeys(gstools_mdls.keys())
            
            for ind, model in enumerate(gstools_mdls):
                fitted_vario = gstools_mdls[model](dim=3, **Cof[model])
            
                kriger = OrdinaryKriging3D(x=X0[:,0], y=X0[:,1], z=X0[:,2], val = Y0,\
                                               variogram_model = fitted_vario, \
                                            exact_values = True)
                y1, error = kriger.execute(style = 'points', xpoints = x1[:,0], ypoints = x1[:,1], zpoints = x1[:, 2])
            
                full_map = np.zeros(sphere_mesh.n_points)
                full_map[kv[:,0].astype(int)] = mapvals[kv[:,0].astype(int)]
                full_map[uv[:,0].astype(int)] = y1
            
                preds[model] = full_map
                y_true = test
                y_pred = preds[model][uv[:,0].astype(int)]
                error = metrics.metric(actual=y_true, interpolated=y_pred, method="rmse")
                nrmse = error/np.std(y_true)
        
                Nrmse[mi, ind] = nrmse
                
        NRMSE[pct] = Nrmse
        
    joblib.dump((ran, NRMSE), projdir + f"/results/rev_many_grf_nrmse_{ran}mm.pkl")
```
