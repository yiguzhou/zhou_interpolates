```python
import joblib

all_sampling = joblib.load(projdir + "/data/sampling/fsLR-4k_reduced-midthickness_samp-dsk33_pct-10_iters-100.pkl")[1]
```

```python
src_polydata = joblib.load(projdir + "/data/fsLR-4k_reduced-midthickness_neuromaps.pkl")
trg_polydata = src_polydata
```

```python
import run_wrappers

for curr_iter in range(10):

    all_known_coords, all_known_verts, all_unknown_coords, all_unknown_verts = all_sampling
    known_coords = all_known_coords[:,:,curr_iter]
    known_verts = all_known_verts[:,curr_iter].astype(int)
    unknown_coords = all_unknown_coords[:,:,curr_iter]
    unknown_verts = all_unknown_verts[:,curr_iter].astype(int)
    
    Yk0, yk1 = run_wrappers.rev_generate_covariates(curr_map="abagen_genepc1", 
                                         all_sampling=(known_coords, known_verts, unknown_coords, unknown_verts), 
                                         src_polydata=src_polydata, 
                                         trg_polydata=trg_polydata, 
                                         nspins=3000, 
                                         datasource='neuromaps')
```
